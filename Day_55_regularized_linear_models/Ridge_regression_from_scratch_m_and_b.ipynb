```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import r2_score
```

```python
X,y=make_regression(n_samples=100,n_features=1,noise=20,random_state=13,n_informative=1,n_targets=1)
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
lr.fit(X,y)
lr.predict(X)
print(r2_score(y,lr.predict(X)))
print(lr.coef_)
print(lr.intercept_)
```

```python
from sklearn.linear_model import Ridge
```

```python
rr=Ridge(alpha=10)
rr.fit(X,y)
rr.predict(X)
print(r2_score(y,rr.predict(X)))
print(rr.coef_)
print(rr.intercept_)
```

```python
rr1 = Ridge(alpha=100)
rr1.fit(X,y)
print(rr1.coef_)
print(rr1.intercept_)
```

```python
plt.plot(X,y,'b.')
plt.plot(X,lr.predict(X),color='red',label='alpha=0')
plt.plot(X,rr.predict(X),color='green',label='alpha=20')
plt.plot(X,rr1.predict(X),color='orange',label='alpha=200')
```

```python
def Linear_regression(X,y,alpha=1):
    X_mean=X.mean()
    y_mean=y.mean()

    num=0
    den=0

    for i in range(X.shape[0]):
        num=num+(y[i]-y_mean)(X[i]-X_mean)
        den=den+(X[i]-X_mean)(X[i]-X_mean)
    m=num/(den+alpha)
    b=y_mean-m*X_mean

    return m,b

```
