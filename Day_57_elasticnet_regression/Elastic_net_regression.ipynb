```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.metrics import r2_score
```

```python
X,y=load_diabetes(return_X_y=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=2)
```

```python
#linear regression
lr=LinearRegression()
lr.fit(X_train,y_train)
y_pred=lr.predict(X_test)
print("R2 score for Linear Regression:",r2_score(y_test,y_pred))
```

```python
#ridge regression
ridge=Ridge(alpha=0.1)
ridge.fit(X_train,y_train)
y_pred_ridge=ridge.predict(X_test)
print("R2 score for Ridge Regression:",r2_score(y_test,y_pred_ridge))
```

```python
#lasso regression
lasso=Lasso(alpha=0.01)
lasso.fit(X_train,y_train)
y_pred_lasso=lasso.predict(X_test)
print("R2 score for Lasso Regression:",r2_score(y_test,y_pred_lasso))
```

```python
#ElasticNet regression
elastic_net=ElasticNet(alpha=0.005, l1_ratio=0.9)
elastic_net.fit(X_train,y_train)
y_pred_elastic_net=elastic_net.predict(X_test)
print("R2 score for ElasticNet Regression:",r2_score(y_test,y_pred_elastic_net))
```
