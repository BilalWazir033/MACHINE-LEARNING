## Mera MLR

```python
class Mera_MLR:
    def __init__(self):
        self.coef_=None
        self.intercept_=None
    def init(self,X_train,y_train):
        X_train=np.insert(X_train,0,1,axis=1) # in X_train we add a new colmn at index 0  of value 1

        betas=np.linalg.inv(np.dot(X_train.T,X_train)).dot(X_train.T).dot(y_train)
        self.intercept_=betas[0]
        self.coef_=betas[1:]


    def predict(self,X_test):
        y_predict=np.dot(X_test,self.coef_) + self.intercept_
        
```

```python
mlr=Mera_MLR()
```

```python
mlr.init(X_train,y_train)
```

```python
y_predict1=mlr.predict(X_test)
```

```python
r2_score(y_test,y_predict1)
```
